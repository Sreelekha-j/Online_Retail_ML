# Phase 5: Ensemble Learning

In Phase 4, Logistic Regression was my best model with ROC-AUC 0.789.

Now I want to see if ensemble models can do better. An ensemble uses many models together instead of one. It is like asking a group of people instead of just one person.

I try three types:
- Random Forest: many trees learn on their own and then vote.
- XGBoost: trees are made one by one, and each new tree fixes the mistakes of the last ones.
- Stacking: I combine different models, and one final model decides how much to trust each of them.

I use the same train and test data as Phase 4, so the comparison is fair. At the end I compare all models and choose the best one.

# Importing all the required Packages

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import joblib
import shap

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, StackingClassifier
from xgboost import XGBClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score,
                             ConfusionMatrixDisplay, RocCurveDisplay)
from sklearn.tree import DecisionTreeClassifier
```

# Loading the data

```python
churn = pd.read_parquet("../data/processed/churn_train.parquet")
print(churn.shape)
```

# Separating X and Y variables

```python
X = churn.drop(columns=['Customer ID', 'Churn'])
```

```python
X.info()
```

```python
y = churn['Churn']
```

```python
y.info()
```

# Splitting into training and test data

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
print(X_train.shape, X_test.shape)
```

```python
print(y_train.mean() , y_test.mean())
```

## Function to evaluate models
- Instead of copying the same metric code for every model, I write a function that does it once.
- It gives back all the scores for a model as one row, so I can put all models in one table and compare them.
- It also gives the ROC-AUC on the training data, so I can check if a model is overfitting.

```python
def evaluate(name, model):
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]
    train_prob = model.predict_proba(X_train)[:, 1]

    return {
        'Model': name,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1': f1_score(y_test, y_pred),
        'ROC_AUC': roc_auc_score(y_test, y_prob),
        'Train_ROC_AUC': roc_auc_score(y_train, train_prob),
    }
```

## Rebuilding the baseline models
- I train Logistic Regression and the depth-5 Decision Tree again with the same settings as Phase 4.
- Because I use the same split, they give the same scores as before, so I can compare them fairly with the ensembles.

```python
log_reg = Pipeline([
    ('log', FunctionTransformer(np.log1p)),
    ('scale', StandardScaler()),
    ('model', LogisticRegression(max_iter=1000)),
])
log_reg.fit(X_train, y_train)

dec_tree_limit = DecisionTreeClassifier(max_depth=5, random_state=42)
dec_tree_limit.fit(X_train, y_train)
```

```python
results = []
results.append(evaluate('Logistic Regression', log_reg))
results.append(evaluate('Decision Tree (depth 5)', dec_tree_limit))
pd.DataFrame(results).round(3)
```

### Insights:
- I used the same split as Phase 4 (random_state = 42), so all models are tested on the same 1,052 customers.
- The baseline models give exactly the same scores as in Phase 4 (Logistic Regression 0.789, Decision Tree 0.769), so the comparison will be fair.

## Random Forest (bagging)
- Random Forest builds many decision trees. Each tree learns from a random sample of customers and features.
- All trees vote, and the final prediction is the average of the votes.
- Because each tree makes different mistakes, averaging them makes the model more stable than one tree.

```python
rf = RandomForestClassifier(n_estimators=300, min_samples_leaf=5, random_state=42, n_jobs=-1)
```

```python
rf.fit(X_train, y_train)
```

```python
results.append(evaluate('Random Forest', rf))
pd.DataFrame(results).round(3)
```

### Insights:
- Random Forest has test ROC-AUC 0.788, almost the same as Logistic Regression (0.789), and much better than the depth-5 tree (0.769).
- It has slightly higher precision (0.76 vs 0.75), but lower recall (0.778 vs 0.795), so it misses a few more churners than Logistic Regression.
- Its train ROC-AUC (0.961) is much higher than its test score (0.788). The trees learn the training data very closely, but the test score is still good, so it is not overfitting badly like the tree with no limit (1.0 vs 0.641).
- So far, the ensemble did not beat the simple Logistic Regression. This means the churn pattern in my data is fairly simple.

## XGBoost (boosting)
- XGBoost builds small trees one after another.
- Each new tree tries to fix the mistakes of the trees before it.
- Adding all the trees together gives the final prediction.

```python
xgb = XGBClassifier(n_estimators=300, learning_rate=0.05, max_depth=3, subsample=0.8, colsample_bytree=0.8, random_state=42, eval_metric='logloss') #creating the model
```

```python
xgb.fit(X_train, y_train) # Training the model
```

```python
results.append(evaluate('XGBoost', xgb)) # adding to the result table
pd.DataFrame(results).round(3)
```

### Insights:
- XGBoost has the highest test ROC-AUC so far (0.792), a little better than Logistic Regression (0.789) and Random Forest (0.788).
- But its accuracy (0.722) and F1 (0.761) are lower than Logistic Regression. ROC-AUC checks how well the model ranks customers, but accuracy and F1 use a fixed cut-off of 0.5. So XGBoost ranks customers well, but 0.5 may not be the best cut-off for it. I will look at this in Phase 6.
- Its train ROC-AUC is 0.874, so the gap between train and test (0.08) is much smaller than Random Forest (0.17). Small trees and a low learning rate helped it overfit less.
- All the models are very close (0.788 to 0.792), so the difference between them is small.

## Stacking
- Stacking combines different models: Logistic Regression, Random Forest and XGBoost.
- Each model predicts a churn probability, and a final Logistic Regression learns how much to trust each one.
- If the models make different mistakes, combining them can be better than any single model.

```python
stack = StackingClassifier(
    estimators=[
        ('lr', log_reg),
        ('rf', rf),
        ('xgb', xgb),
    ],
    final_estimator=LogisticRegression(max_iter=1000),
    cv=5,
    n_jobs=-1,
)
```

```python
stack.fit(X_train, y_train)
```

```python
results.append(evaluate('Stacking', stack)) #: add it to the table
pd.DataFrame(results).round(3)
```

```python
stack.final_estimator_.coef_ #how much the final model trusts each base model (numbers in the order lr, rf, xgb):
```

### Insights:
- Stacking has the highest test ROC-AUC (0.796), better than XGBoost (0.792) and Logistic Regression (0.789).
- It also has good accuracy (0.736) and F1 (0.771), almost the same as Logistic Regression, so it is the most balanced model so far.
- The final model trusts XGBoost the most (2.00), then Logistic Regression (1.58), then Random Forest (1.40).
- Combining the models helped a little, because Logistic Regression and the tree models make slightly different mistakes.
- The improvement is small (0.796 vs 0.789), so I will check with cross-validation if it is really better.

## Cross-validation
- All my models have very close test scores, so I use 5-fold cross-validation to check which one is really better.
- I look at the average score (how good) and the standard deviation (how stable).

```python
cv_lr = cross_val_score(log_reg, X_train, y_train, cv=5, scoring='roc_auc')
cv_rf = cross_val_score(rf, X_train, y_train, cv=5, scoring='roc_auc')
cv_xgb = cross_val_score(xgb, X_train, y_train, cv=5, scoring='roc_auc')
cv_stack = cross_val_score(stack, X_train, y_train, cv=5, scoring='roc_auc')
```

```python
print("Logistic Regression:", round(cv_lr.mean(), 3), "±", round(cv_lr.std(), 3))
print("Random Forest:", round(cv_rf.mean(), 3), "±", round(cv_rf.std(), 3))
print("XGBoost:", round(cv_xgb.mean(), 3), "±", round(cv_xgb.std(), 3))
print("Stacking:", round(cv_stack.mean(), 3), "±", round(cv_stack.std(), 3))
```

### Insights:
- Stacking has the highest average cross-validation ROC-AUC (0.808), then XGBoost (0.805), Random Forest (0.803) and Logistic Regression (0.801).
- This is the same order as the test scores, so Stacking being the best is not just luck from one test set.
- XGBoost is the most stable, because it has the smallest standard deviation (± 0.005). Logistic Regression is the least stable (± 0.011).
- The difference between the best and worst model is only 0.007, so all models are very close. The ensembles improved a little over Logistic Regression, but not by much.

### Comparing all models
- I put all models in one table with all the metrics and the cross-validation scores.
- I also draw the ROC curves of all models together and look at the confusion matrix of the best model.

```python
results_df = pd.DataFrame(results)

cv_means = {
    'Logistic Regression': cv_lr.mean(),
    'Random Forest': cv_rf.mean(),
    'XGBoost': cv_xgb.mean(),
    'Stacking': cv_stack.mean(),
}
results_df['CV_ROC_AUC'] = results_df['Model'].map(cv_means)

results_df.sort_values('ROC_AUC', ascending=False).round(3)
```

### plotting ROC Curves for all models

```python
fig, ax = plt.subplots(figsize=(8, 6))

RocCurveDisplay.from_estimator(stack, X_test, y_test, ax=ax, name='Stacking')
RocCurveDisplay.from_estimator(xgb, X_test, y_test, ax=ax, name='XGBoost')
RocCurveDisplay.from_estimator(log_reg, X_test, y_test, ax=ax, name='Logistic Regression')
RocCurveDisplay.from_estimator(rf, X_test, y_test, ax=ax, name='Random Forest')
RocCurveDisplay.from_estimator(dec_tree_limit, X_test, y_test, ax=ax, name='Decision Tree (depth 5)')

ax.plot([0, 1], [0, 1], 'k--', label='Random guess')
ax.set_title('ROC curves: all models', fontweight='bold')
ax.legend()
plt.show()
```

### Confusion matrix of stacking

```python
ConfusionMatrixDisplay.from_estimator(stack, X_test, y_test,
                                      display_labels=['Stayed', 'Churned'])
plt.title('Confusion matrix: Stacking', fontweight='bold')
plt.show()
```

### Insights:
- Stacking is at the top of the table, with the best test ROC-AUC (0.796) and the best cross-validation ROC-AUC (0.808).
- The ROC curves of all models are very close, which shows again that the models are similar. The depth-5 tree is a little lower than the others.
- In the confusion matrix, Stacking missed 127 churners and gave 151 false alarms. Logistic Regression missed 122 churners and gave 157 false alarms.
- So Stacking gives fewer false alarms, but it misses a few more churners at the 0.5 cut-off. Its recall (0.786) is a little lower than Logistic Regression (0.795).

## Model selection
- I chose Stacking as my best model.
- It has the highest test ROC-AUC (0.796) and the highest cross-validation ROC-AUC (0.808 ± 0.007), so it is the best at telling churners and non-churners apart, and this is not just luck from one test set.
- It also has the best accuracy (0.736) and almost the same F1 as Logistic Regression (0.771 vs 0.772).
- Its recall (0.786) is a little lower than Logistic Regression (0.795), so at the 0.5 cut-off it misses 5 more churners. Because recall matters most for the business, I will fix this in Phase 6 by choosing a better cut-off than 0.5. A model that ranks customers better (higher ROC-AUC) can reach higher recall with a better cut-off.
- Its train ROC-AUC (0.891) is higher than its test score (0.796), but the test score is still the best of all models, so it is not overfitting badly.
- The downside is that Stacking is more complex and slower to train than Logistic Regression. But the improvement was the same on both the test set and cross-validation, so I think it is worth it.
- All models were very close (0.788 to 0.796), so the churn pattern in my data is fairly simple, and the ensembles only improved a little over Logistic Regression.
